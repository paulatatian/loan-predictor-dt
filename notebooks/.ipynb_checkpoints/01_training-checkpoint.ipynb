# 🏦 Predictor de Aprobación de Préstamos

### Algoritmo: Árbol de Decisión

**¿Qué hace este notebook?**
Entrenamos un modelo de Machine Learning que predice si un banco debe aprobar o rechazar un préstamo,
basándose en características del solicitante como ingresos, historial crediticio, etc.

**Dataset:** Loan Approval Dataset
- Fuente: Kaggle (https://www.kaggle.com/datasets/altruistdelhite04/loan-prediction-problem-dataset)
- 614 registros reales
- 12 variables (features)

## Paso 1: Importar librerías
Las librerías son herramientas que ya vienen programadas y nos ahorran escribir código desde cero.

```python
# Manejo de datos
import pandas as pd       # Para leer y manipular tablas de datos
import numpy as np        # Para operaciones matemáticas

# Visualizaciones
import matplotlib.pyplot as plt   # Para crear gráficos
import seaborn as sns             # Para gráficos más bonitos

# Machine Learning
from sklearn.tree import DecisionTreeClassifier, plot_tree  # Árbol de Decisión
from sklearn.model_selection import train_test_split         # Dividir datos
from sklearn.preprocessing import LabelEncoder               # Convertir texto a números
from sklearn.metrics import (accuracy_score, precision_score, 
                             recall_score, f1_score, 
                             confusion_matrix, classification_report)  # Métricas

# Guardar el modelo
import joblib

# Para no ver advertencias innecesarias
import warnings
warnings.filterwarnings('ignore')

print('✅ Librerías importadas correctamente')
```

## Paso 2: Cargar los datos

**¿Cómo obtener el dataset?**
1. Ve a: https://www.kaggle.com/datasets/altruistdelhite04/loan-prediction-problem-dataset
2. Descarga `train.csv`
3. Colócalo en la carpeta `data/raw/` con el nombre `loan_data.csv`

**Descripción de las columnas:**
- `Loan_ID`: Identificador único del préstamo
- `Gender`: Género del solicitante (Male/Female)
- `Married`: Estado civil (Yes/No)
- `Dependents`: Número de dependientes (0, 1, 2, 3+)
- `Education`: Nivel educativo (Graduate/Not Graduate)
- `Self_Employed`: ¿Es autoempleado? (Yes/No)
- `ApplicantIncome`: Ingresos del solicitante
- `CoapplicantIncome`: Ingresos del co-solicitante
- `LoanAmount`: Monto del préstamo (en miles)
- `Loan_Amount_Term`: Plazo del préstamo (en meses)
- `Credit_History`: Historial crediticio (1=bueno, 0=malo)
- `Property_Area`: Área de la propiedad (Urban/Semiurban/Rural)
- `Loan_Status`: ¿Fue aprobado? (Y=Sí, N=No) ← Esta es la que queremos predecir

```python
# Cargar el dataset
df = pd.read_csv('../data/raw/loan_data.csv')

# Ver las primeras 5 filas
print(f'📊 El dataset tiene {df.shape[0]} filas y {df.shape[1]} columnas')
print('\nPrimeras 5 filas:')
df.head()
```

```python
# Información general del dataset
print('Información del dataset:')
df.info()
```

```python
# Estadísticas descriptivas de las columnas numéricas
print('Estadísticas descriptivas:')
df.describe()
```

## Paso 3: Análisis Exploratorio de Datos (EDA)
Antes de entrenar el modelo, necesitamos entender bien nuestros datos.

```python
# ¿Cuántos préstamos fueron aprobados vs rechazados?
print('Distribución de Loan_Status (variable objetivo):')
print(df['Loan_Status'].value_counts())
print(f'\nPorcentaje aprobados: {df["Loan_Status"].value_counts(normalize=True)["Y"]*100:.1f}%')
print(f'Porcentaje rechazados: {df["Loan_Status"].value_counts(normalize=True)["N"]*100:.1f}%')

# Gráfico de la distribución
plt.figure(figsize=(6, 4))
colors = ['#2ecc71', '#e74c3c']
df['Loan_Status'].value_counts().plot(kind='bar', color=colors)
plt.title('Distribución de Aprobación de Préstamos', fontsize=14, fontweight='bold')
plt.xlabel('Estado del Préstamo (Y=Aprobado, N=Rechazado)')
plt.ylabel('Cantidad')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('../docs/distribucion_target.png', dpi=150)
plt.show()
print('✅ Gráfico guardado')
```

```python
# Valores nulos por columna
nulos = df.isnull().sum()
nulos_porcentaje = (df.isnull().sum() / len(df)) * 100

tabla_nulos = pd.DataFrame({
    'Valores Nulos': nulos,
    'Porcentaje (%)': nulos_porcentaje.round(2)
})

print('Valores nulos por columna:')
print(tabla_nulos[tabla_nulos['Valores Nulos'] > 0])
```

```python
# Análisis visual de variables categóricas vs aprobación
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
fig.suptitle('Relación entre variables categóricas y aprobación de préstamo', 
             fontsize=14, fontweight='bold')

categoricas = ['Gender', 'Married', 'Education', 'Self_Employed', 'Property_Area', 'Credit_History']

for i, col in enumerate(categoricas):
    ax = axes[i//3][i%3]
    pd.crosstab(df[col], df['Loan_Status']).plot(kind='bar', ax=ax, 
                                                   color=['#e74c3c', '#2ecc71'])
    ax.set_title(f'{col} vs Loan_Status')
    ax.set_xlabel(col)
    ax.set_ylabel('Cantidad')
    ax.tick_params(axis='x', rotation=45)
    ax.legend(['Rechazado', 'Aprobado'])

plt.tight_layout()
plt.savefig('../docs/variables_categoricas.png', dpi=150)
plt.show()
```

```python
# Distribución de variables numéricas
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

numericas = ['ApplicantIncome', 'LoanAmount', 'Loan_Amount_Term']
for i, col in enumerate(numericas):
    axes[i].hist(df[col].dropna(), bins=30, color='#3498db', edgecolor='white')
    axes[i].set_title(f'Distribución de {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Frecuencia')

plt.tight_layout()
plt.savefig('../docs/variables_numericas.png', dpi=150)
plt.show()
```

## Paso 4: Preprocesamiento de datos
Los modelos de ML solo entienden números, no texto. Debemos:
1. Eliminar columnas que no aportan información
2. Rellenar valores nulos
3. Convertir texto a números

```python
# Trabajamos con una copia para no modificar el original
df_clean = df.copy()

# 1. Eliminar columna Loan_ID (solo es un identificador, no aporta)
df_clean = df_clean.drop('Loan_ID', axis=1)
print('✅ Columna Loan_ID eliminada')

# 2. Rellenar valores nulos
# Para columnas de texto: usamos la moda (valor más frecuente)
columnas_texto = ['Gender', 'Married', 'Dependents', 'Self_Employed', 'Credit_History']
for col in columnas_texto:
    moda = df_clean[col].mode()[0]
    df_clean[col].fillna(moda, inplace=True)
    print(f'  → {col}: valores nulos rellenados con "{moda}"')

# Para columnas numéricas: usamos la mediana (más robusta que el promedio)
df_clean['LoanAmount'].fillna(df_clean['LoanAmount'].median(), inplace=True)
df_clean['Loan_Amount_Term'].fillna(df_clean['Loan_Amount_Term'].median(), inplace=True)
print('  → LoanAmount y Loan_Amount_Term: rellenados con la mediana')

print(f'\n✅ Valores nulos restantes: {df_clean.isnull().sum().sum()}')
```

```python
# 3. Convertir variables de texto a números usando LabelEncoder
# LabelEncoder asigna un número a cada categoría: e.g., Male=1, Female=0

le = LabelEncoder()

columnas_categoricas = ['Gender', 'Married', 'Dependents', 'Education', 
                         'Self_Employed', 'Property_Area', 'Loan_Status']

# Guardamos los mapeos para referencia
mapeos = {}
for col in columnas_categoricas:
    df_clean[col] = le.fit_transform(df_clean[col])
    mapeos[col] = dict(zip(le.classes_, le.transform(le.classes_)))
    print(f'  → {col}: {mapeos[col]}')

print('\n✅ Variables categóricas convertidas a números')
```

```python
# 4. Separar features (X) y variable objetivo (y)
# X = todo lo que el modelo usa para aprender
# y = lo que queremos predecir

X = df_clean.drop('Loan_Status', axis=1)  # Todo menos la columna objetivo
y = df_clean['Loan_Status']               # Solo la columna objetivo

print(f'Features (X): {X.shape[1]} columnas, {X.shape[0]} filas')
print(f'Features usadas: {list(X.columns)}')
print(f'\nTarget (y): {y.shape[0]} valores')
print(f'Distribución: {y.value_counts().to_dict()} (1=Aprobado, 0=Rechazado)')
```

```python
# 5. Dividir en conjunto de entrenamiento (80%) y prueba (20%)
# El modelo aprende con el 80% y lo evaluamos con el 20% que no vio

X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.2,      # 20% para prueba
    random_state=42,    # Para reproducibilidad (siempre el mismo split)
    stratify=y          # Mantener proporción de clases
)

print(f'Datos de entrenamiento: {X_train.shape[0]} registros')
print(f'Datos de prueba:        {X_test.shape[0]} registros')
print('\n✅ División completada')
```

## Paso 5: Entrenamiento del Árbol de Decisión

**¿Cómo funciona un Árbol de Decisión?**
Es como un juego de preguntas de sí/no. El modelo aprende cuáles preguntas hacer
y en qué orden, para llegar a la mejor decisión final.

Ejemplo: ¿Tiene buen historial crediticio? → Si NO → Rechazar. Si SÍ → ¿Cuánto gana? → ...

```python
# Crear el modelo Árbol de Decisión
model = DecisionTreeClassifier(
    max_depth=5,          # Profundidad máxima del árbol (evita sobreajuste)
    min_samples_split=10, # Mínimo de muestras para dividir un nodo
    min_samples_leaf=5,   # Mínimo de muestras en una hoja
    random_state=42       # Para reproducibilidad
)

# Entrenar el modelo con los datos de entrenamiento
# .fit() = 'aprende' de los datos
model.fit(X_train, y_train)

print('✅ Modelo entrenado exitosamente')
print(f'Profundidad real del árbol: {model.get_depth()}')
print(f'Número de hojas: {model.get_n_leaves()}')
```

```python
# Visualizar el árbol de decisión

plt.figure(figsize=(20, 10))

plot_tree(
    model,
    feature_names=list(X.columns),
    class_names=['Rechazado', 'Aprobado'],
    filled=True,
    rounded=True,
    fontsize=8
)

plt.title(
    'Árbol de Decisión - Aprobación de Préstamos',
    fontsize=16,
    fontweight='bold'
)

plt.tight_layout()

plt.savefig(
    '../docs/arbol_decision.png',
    dpi=100,
    bbox_inches='tight'
)

plt.show()

print('✅ Árbol guardado como imagen')
```

## Paso 6: Evaluación del modelo
Verificamos qué tan bien predice el modelo con datos que NO vio durante el entrenamiento.

```python
# Hacer predicciones con el conjunto de prueba
y_pred = model.predict(X_test)

# Calcular métricas
accuracy  = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall    = recall_score(y_test, y_pred)
f1        = f1_score(y_test, y_pred)

print('=' * 45)
print('          MÉTRICAS DEL MODELO')
print('=' * 45)
print(f'  Accuracy  (exactitud):  {accuracy:.4f}  ({accuracy*100:.1f}%)')
print(f'  Precision:              {precision:.4f}  ({precision*100:.1f}%)')
print(f'  Recall    (sensibilidad):{recall:.4f}  ({recall*100:.1f}%)')
print(f'  F1-Score:               {f1:.4f}  ({f1*100:.1f}%)')
print('=' * 45)
print()
print('¿Qué significan estas métricas?')
print(f'  → El modelo acierta el {accuracy*100:.1f}% de las veces en total')
print(f'  → Cuando predice "Aprobado", acierta el {precision*100:.1f}% de las veces')
print(f'  → Detecta el {recall*100:.1f}% de los préstamos que realmente deben aprobarse')
```

```python
# Matriz de confusión
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Rechazado', 'Aprobado'],
            yticklabels=['Rechazado', 'Aprobado'],
            linewidths=0.5)
plt.title('Matriz de Confusión', fontsize=14, fontweight='bold')
plt.ylabel('Valor Real', fontsize=12)
plt.xlabel('Predicción del Modelo', fontsize=12)
plt.tight_layout()
plt.savefig('../docs/matriz_confusion.png', dpi=150)
plt.show()

print('Interpretación de la matriz:')
print(f'  Verdaderos Negativos (Rechazado correcto): {cm[0][0]}')
print(f'  Falsos Positivos (Aprobó cuando debía rechazar): {cm[0][1]}')
print(f'  Falsos Negativos (Rechazó cuando debía aprobar): {cm[1][0]}')
print(f'  Verdaderos Positivos (Aprobado correcto): {cm[1][1]}')
```

```python
# Importancia de cada variable en la decisión del modelo
importancias = pd.DataFrame({
    'Feature': X.columns,
    'Importancia': model.feature_importances_
}).sort_values('Importancia', ascending=False)

plt.figure(figsize=(10, 6))
bars = plt.barh(importancias['Feature'], importancias['Importancia'], 
                color='#3498db', edgecolor='white')
plt.xlabel('Importancia relativa', fontsize=12)
plt.title('¿Qué variables influyen más en la decisión?', fontsize=14, fontweight='bold')
plt.gca().invert_yaxis()
for bar, val in zip(bars, importancias['Importancia']):
    plt.text(bar.get_width() + 0.001, bar.get_y() + bar.get_height()/2,
             f'{val:.3f}', va='center', fontsize=9)
plt.tight_layout()
plt.savefig('../docs/feature_importance.png', dpi=150)
plt.show()

print('Variable más importante:', importancias.iloc[0]['Feature'],
      f"({importancias.iloc[0]['Importancia']*100:.1f}% de influencia)")
```

```python
# Reporte completo de clasificación
print('Reporte completo:')
print(classification_report(y_test, y_pred, 
                            target_names=['Rechazado', 'Aprobado']))
```

## Paso 7: Serializar (guardar) el modelo
Guardamos el modelo entrenado para poder usarlo en la aplicación web sin tener que re-entrenarlo cada vez.

```python
import os
os.makedirs('../models', exist_ok=True)

# Guardar el modelo
joblib.dump(model, '../models/loan_model.pkl')

# Guardar la lista de features para referencia en la app
feature_names = list(X.columns)
joblib.dump(feature_names, '../models/feature_names.pkl')

# Guardar los mapeos de categorías (útil para la app)
joblib.dump(mapeos, '../models/label_mapeos.pkl')

print('✅ Modelo guardado en: models/loan_model.pkl')
print('✅ Nombres de features en: models/feature_names.pkl')
print('✅ Mapeos de categorías en: models/label_mapeos.pkl')

# Verificar que se puede cargar
modelo_cargado = joblib.load('../models/loan_model.pkl')
prueba = modelo_cargado.predict([X_test.iloc[0]])
print(f'\n✅ Verificación: modelo cargado y funcionando. Predicción de prueba: {"Aprobado" if prueba[0] == 1 else "Rechazado"}')
```
